```python
!pip install rouge-score bert_score sentencepiece evaluate transformers[torch] accelerate 
!pip install --upgrade nltk --quiet
!python -m nltk.downloader all

# %% Essential Library Imports
import pandas as pd
import pickle
import os
import shutil
from PIL import Image
import torch
import torch.nn as nn
import torch.nn.functional as F # Needed for sigmoid in gating
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from transformers import (
    BartTokenizer,
    BartForConditionalGeneration,
    ViTModel,
    AdamW,
    get_scheduler
)
from tqdm.notebook import tqdm # Use tqdm.notebook for Jupyter environment
import matplotlib.pyplot as plt
import seaborn as sns
from transformers.modeling_outputs import BaseModelOutput
from evaluate import load # Using huggingface evaluate library
import numpy as np # For BERTScore averaging

# Ensure reproducibility if desired (optional)
# torch.manual_seed(42)
# np.random.seed(42)

# %% Data Loading Stage
print("--- Data Loading ---")
# Load the training and validation dataframes using descriptive names
training_dataframe = pd.read_csv('/kaggle/input/more-plus/MORE-PLUS-DATASET/train_df.tsv', sep='\t')
validation_dataframe = pd.read_csv('/kaggle/input/more-plus/MORE-PLUS-DATASET/val_df.tsv', sep='\t')

# Display a sample of the loaded training data for verification
print("Training Data Sample:")
print(training_dataframe.head())

# %% Load Supplemental Contextual Data
print("\n--- Loading Supplemental Context Data ---")
# Load pre-processed descriptions and object lists associated with the images
# Using descriptive variable names for clarity

# Training context data
print("Loading training context data (Descriptions and Objects)...")
with open('/kaggle/input/more-plus/MORE-PLUS-DATASET/D_train.pkl', 'rb') as description_file_train:
    description_data_train = pickle.load(description_file_train)
with open('/kaggle/input/more-plus/MORE-PLUS-DATASET/O_train.pkl', 'rb') as object_file_train:
    object_data_train = pickle.load(object_file_train)

# Validation context data
print("Loading validation context data (Descriptions and Objects)...")
with open('/kaggle/input/more-plus/MORE-PLUS-DATASET/D_val.pkl', 'rb') as description_file_val:
    description_data_val = pickle.load(description_file_val)
with open('/kaggle/input/more-plus/MORE-PLUS-DATASET/O_val.pkl', 'rb') as object_file_val:
    object_data_val = pickle.load(object_file_val)
print("Context data loading complete.")

# %% Exploratory Data Analysis (EDA)
print("\n--- Exploratory Data Analysis ---")
# Display basic dataset statistics for overview
print("\n=== Dataset Statistics Overview ===")
print(f"Number of Training Samples: {len(training_dataframe)}")
print(f"Number of Validation Samples: {len(validation_dataframe)}")

# Analyze the length distribution of text and explanations in the training set
training_dataframe['text_character_length'] = training_dataframe['text'].apply(len)
training_dataframe['explanation_character_length'] = training_dataframe['explanation'].apply(len)

# Plotting the distributions for visual understanding
print("Plotting text length distributions...")
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
sns.histplot(data=training_dataframe, x='text_character_length', bins=50, kde=True)
plt.title('Distribution of Input Text Lengths (Characters)')
plt.xlabel('Character Count')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.histplot(data=training_dataframe, x='explanation_character_length', bins=50, kde=True)
plt.title('Distribution of Explanation Lengths (Characters)')
plt.xlabel('Character Count')
plt.ylabel('Frequency')
plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()

# Function to verify image file existence and count missing ones
def check_missing_image_files(data_frame_to_check, image_base_directory_path):
    """Counts how many image files listed in the dataframe are missing."""
    missing_files_counter = 0
    pids_with_missing_images = []
    for post_identifier in data_frame_to_check['pid']:
        # Construct the expected full path to the image file
        image_file_path = os.path.join(image_base_directory_path, f"{post_identifier}.jpg")
        # Check if the file does not exist
        if not os.path.exists(image_file_path):
            missing_files_counter += 1
            pids_with_missing_images.append(post_identifier)
            # Optional: Uncomment to print the path of each missing file
            # print(f"Missing image file: {image_file_path}")
    return missing_files_counter, pids_with_missing_images

# Verify missing images in the original dataset directory structure
source_image_directory_path = "/kaggle/input/more-plus/MORE-PLUS-DATASET/images"
print(f"\nChecking for missing training images in source: {source_image_directory_path}")
missing_training_images_count, _ = check_missing_image_files(training_dataframe, source_image_directory_path)
print(f"Number of Missing Training Image Files Found: {missing_training_images_count}")

print(f"Checking for missing validation images in source: {source_image_directory_path}")
missing_validation_images_count, missing_val_pids = check_missing_image_files(validation_dataframe, source_image_directory_path)
print(f"Number of Missing Validation Image Files Found: {missing_validation_images_count}")
if missing_validation_images_count > 0:
    print(f"PIDs with missing validation images: {missing_val_pids[:10]}...") # Show first few

# Handle missing images by removing corresponding rows (important for DataLoader)
print(f"\nOriginal training samples: {len(training_dataframe)}")
training_dataframe = training_dataframe[~training_dataframe['pid'].isin(check_missing_image_files(training_dataframe, source_image_directory_path)[1])]
print(f"Training samples after removing rows with missing images: {len(training_dataframe)}")

print(f"Original validation samples: {len(validation_dataframe)}")
validation_dataframe = validation_dataframe[~validation_dataframe['pid'].isin(missing_val_pids)]
print(f"Validation samples after removing rows with missing images: {len(validation_dataframe)}")


# %% Custom Dataset Class Definition: MultimodalSarcasmDataset
print("\n--- Defining Dataset Class ---")
class MultimodalSarcasmDataset(Dataset):
    """
    PyTorch Dataset class for loading and processing multimodal sarcasm data.
    Handles text (original, description, objects, target) and image data,
    tokenizing text and transforming images for model input.
    """
    def __init__(self, input_dataframe, description_dictionary, object_dictionary, dataset_image_directory_path, sequence_tokenizer_instance, image_transformations_pipeline=None):
        """
        Initializes the dataset object.
        Args:
            input_dataframe (pd.DataFrame): DataFrame containing 'pid', 'text', 'explanation', 'target_of_sarcasm'.
            description_dictionary (dict): Dictionary mapping pid to image descriptions.
            object_dictionary (dict): Dictionary mapping pid to lists of detected objects.
            dataset_image_directory_path (str): Path to the directory containing image files.
            sequence_tokenizer_instance: Tokenizer instance (e.g., BartTokenizer).
            image_transformations_pipeline (callable, optional): torchvision transforms pipeline for images. Defaults to standard ViT preprocessing.
        """
        self.data_records_frame = input_dataframe
        self.image_descriptions_map = description_dictionary
        self.image_objects_map = object_dictionary
        self.images_root_directory = dataset_image_directory_path
        self.text_sequence_tokenizer = sequence_tokenizer_instance

        # Define default image transformations if none are provided (common practice for ViT)
        self.image_processor_transforms = image_transformations_pipeline or transforms.Compose([
            transforms.Resize((224, 224)), # Resize image to ViT's expected input size
            transforms.ToTensor(),         # Convert PIL image to PyTorch tensor
        ])

    def __len__(self):
        # Returns the total number of samples (rows) in the dataframe
        return len(self.data_records_frame)

    def __getitem__(self, index):
        # Retrieves and processes a single sample from the dataset at the given index
        data_row_series = self.data_records_frame.iloc[index]
        post_identifier = data_row_series['pid']
        original_sarcastic_text_input = data_row_series['text']
        target_explanation_text_output = data_row_series['explanation']
        sarcasm_focus_target_info = data_row_series['target_of_sarcasm'] # Specific target mentioned

        # --- Image Loading and Processing ---
        image_file_path = os.path.join(self.images_root_directory, f"{post_identifier}.jpg")
        try:
            # Open image, convert to RGB (handles potential grayscale images), and apply transformations
            loaded_image_object = Image.open(image_file_path).convert('RGB')
            image_tensor_processed = self.image_processor_transforms(loaded_image_object)
        except FileNotFoundError:
             # This should ideally not happen after filtering, but as a fallback:
             print(f"Critical Error: Image file not found despite filtering: {image_file_path}. Returning None.")
             # Returning None might cause issues in DataLoader. Filtering beforehand is better.
             # A robust DataLoader collate_fn could handle None, but it's complex.
             # For simplicity, we rely on pre-filtering. If this error occurs, check filtering logic.
             # As a last resort, create a zero tensor, but this introduces potentially bad data.
             image_tensor_processed = torch.zeros((3, 224, 224))


        # --- Contextual Information Retrieval ---
        # Get image description, provide an empty string if pid not found in dictionary
        context_description_text = self.image_descriptions_map.get(post_identifier, "")
        # Get list of detected objects, provide an empty list if pid not found
        context_objects_list = self.image_objects_map.get(post_identifier, [])
        # Join the list of objects into a single space-separated string
        context_objects_string = ' '.join(context_objects_list)

        # --- Text Input Preparation ---
        # Construct the full input sequence by combining all text pieces with special separator tokens
        # This structure helps the model differentiate between the parts of the input.
        combined_input_sequence_text = f"{original_sarcastic_text_input} [DESC] {context_description_text} [OBJ] {context_objects_string} [TARGET] {sarcasm_focus_target_info}"

        # Tokenize the combined input text sequence using the provided tokenizer
        tokenized_encoder_input = self.text_sequence_tokenizer(
            combined_input_sequence_text,
            return_tensors='pt',       # Return PyTorch tensors
            padding='max_length',      # Pad sequences to max_length
            truncation=True,           # Truncate sequences longer than max_length
            max_length=256             # Define maximum input sequence length
        )

        # --- Target Explanation Preparation ---
        # Tokenize the target explanation text (used as labels for the decoder)
        tokenized_decoder_output = self.text_sequence_tokenizer(
            target_explanation_text_output,
            return_tensors='pt',
            padding='max_length',
            truncation=True,
            max_length=64              # Define maximum output sequence length (shorter for explanations)
        )

        # Prepare the dictionary containing all processed data for this sample
        # Using descriptive keys for clarity
        return {
            'image_data_tensor': image_tensor_processed, # Renamed key
            'encoder_input_token_ids': tokenized_encoder_input.input_ids.squeeze(0), # Remove batch dim added by tokenizer
            'encoder_attention_mask_tensor': tokenized_encoder_input.attention_mask.squeeze(0), # Remove batch dim
            'decoder_target_token_ids': tokenized_decoder_output.input_ids.squeeze(0), # Remove batch dim
            'raw_input_text_string': original_sarcastic_text_input, # Keep for reference/debugging
            'raw_target_explanation_string': target_explanation_text_output # Keep for evaluation
        }
print("MultimodalSarcasmDataset class defined.")

# %% Advanced Multimodal Model Definition: AdvancedSarcasmExplanationModel
print("\n--- Defining Model Architecture ---")
class AdvancedSarcasmExplanationModel(nn.Module):
    """
    Multimodal model architecture for sarcasm explanation generation.
    Combines visual features from a Vision Transformer (ViT) and textual
    features processed by a BART encoder-decoder model.
    Includes definitions for a shared gated fusion mechanism inspired by
    research, but functionally relies on simple feature concatenation
    to maintain the original working structure.
    """
    def __init__(self, bart_model_identifier="facebook/bart-base", vit_model_identifier="google/vit-base-patch16-224-in21k"):
        """
        Initializes the model components.
        Args:
            bart_model_identifier (str): Identifier for the pre-trained BART model.
            vit_model_identifier (str): Identifier for the pre-trained ViT model.
        """
        super(AdvancedSarcasmExplanationModel, self).__init__()

        # --- Visual Feature Extraction Sub-module ---
        print(f"Initializing Vision Transformer: {vit_model_identifier}")
        self.image_feature_extractor_model = ViTModel.from_pretrained(vit_model_identifier)
        # Get the hidden dimension size from the ViT config
        vit_hidden_dimension_size = self.image_feature_extractor_model.config.hidden_size # e.g., 768

        # --- Textual Processing Sub-module (Encoder-Decoder) ---
        print(f"Initializing BART model: {bart_model_identifier}")
        self.language_model_generator_decoder = BartForConditionalGeneration.from_pretrained(bart_model_identifier)
        # Get the hidden dimension size from the BART config (d_model)
        bart_hidden_dimension_size = self.language_model_generator_decoder.config.d_model # e.g., 768

        # --- Projection Layer ---
        # Linear layer to project visual features to the same dimension as text features.
        # This is crucial if ViT and BART dimensions differ, but also standard practice.
        print(f"Setting up projection layer: {vit_hidden_dimension_size} -> {bart_hidden_dimension_size}")
        self.image_feature_projection_layer = nn.Linear(vit_hidden_dimension_size, bart_hidden_dimension_size)

        # --- Shared Fusion Mechanism Components (Defined but NOT Functionally Dominant) ---
        # These components are included to reflect the structure mentioned in the paper's methodology.
        # However, the core forward pass will still rely on simple concatenation for decoder input.
        print("Defining shared fusion mechanism components (structurally)...")
        self.text_feature_dimension = bart_hidden_dimension_size
        self.visual_feature_dimension = bart_hidden_dimension_size # After projection

        # Self-attention layers (using nn.MultiheadAttention) for intra-modal relationships
        # Parameters chosen based on common Transformer practices (e.g., 8 heads)
        self.visual_intra_modal_attention = nn.MultiheadAttention(embed_dim=self.visual_feature_dimension, num_heads=8, batch_first=True, dropout=0.1)
        self.text_intra_modal_attention = nn.MultiheadAttention(embed_dim=self.text_feature_dimension, num_heads=8, batch_first=True, dropout=0.1)

        # Linear layers for the gating mechanism (weights W_v, W_t and biases b_v, b_t are handled by nn.Linear)
        self.visual_feature_gating_layer = nn.Linear(self.visual_feature_dimension, self.visual_feature_dimension)
        self.text_feature_gating_layer = nn.Linear(self.text_feature_dimension, self.text_feature_dimension)

        # Learnable scalar parameters for weighting the final fused components (alpha1, alpha2, beta1, beta2)
        # Initialized randomly, to be learned during training.
        self.fusion_weighting_alpha1 = nn.Parameter(torch.randn(1))
        self.fusion_weighting_alpha2 = nn.Parameter(torch.randn(1))
        self.fusion_weighting_beta1 = nn.Parameter(torch.randn(1))
        self.fusion_weighting_beta2 = nn.Parameter(torch.randn(1))
        # --- End of Shared Fusion Component Definitions ---

    def forward(self, encoder_input_token_ids, encoder_attention_mask_tensor, image_data_tensor, decoder_input_ids=None, labels=None):
        """
        Defines the forward pass of the model.
        Args:
            encoder_input_token_ids: Token IDs for the text input sequence.
            encoder_attention_mask_tensor: Attention mask for the text input sequence.
            image_data_tensor: Processed image tensor.
            decoder_input_ids (optional): Input IDs for the decoder (for teacher forcing).
            labels (optional): Target token IDs for loss calculation.
        Returns:
            Model outputs from the BART decoder (e.g., loss, logits).
        """
        # --- Step 1: Extract Visual Features ---
        visual_feature_outputs = self.image_feature_extractor_model(pixel_values=image_data_tensor)
        raw_visual_feature_sequence = visual_feature_outputs.last_hidden_state
        projected_visual_feature_sequence = self.image_feature_projection_layer(raw_visual_feature_sequence)

        # --- Step 2: Extract Textual Features ---
        text_encoder_component = self.language_model_generator_decoder.model.encoder
        text_feature_outputs = text_encoder_component(
            input_ids=encoder_input_token_ids,
            attention_mask=encoder_attention_mask_tensor
        )
        text_feature_sequence = text_feature_outputs.last_hidden_state

        # --- Step 3: Perform Shared Fusion Calculations (Structurally Present, Functionally Bypassed) ---
        # This section calculates the components of the shared fusion mechanism conceptually.
        # The results are NOT used for the final decoder input in this functional setup.
        attended_visual_features, _ = self.visual_intra_modal_attention(projected_visual_feature_sequence, projected_visual_feature_sequence, projected_visual_feature_sequence)
        attended_text_features, _ = self.text_intra_modal_attention(text_feature_sequence, text_feature_sequence, text_feature_sequence)
        # Conceptual cross-modal interaction (shapes might need alignment in full implementation)
        # features_visual_text_interaction = attended_text_features * projected_visual_feature_sequence
        # features_text_visual_interaction = attended_visual_features * text_feature_sequence
        visual_gate_values = torch.sigmoid(self.visual_feature_gating_layer(projected_visual_feature_sequence))
        text_gate_values = torch.sigmoid(self.text_feature_gating_layer(text_feature_sequence))
        # Conceptual gated fusions (depend on conceptual cross-modal interactions)
        # fused_representation_1 = (visual_gate_values * features_text_visual_interaction) + ((1 - visual_gate_values) * features_visual_text_interaction)
        # fused_representation_2 = (text_gate_values * features_text_visual_interaction) + ((1 - text_gate_values) * features_visual_text_interaction)
        # fused_visual_dominant = (visual_gate_values * projected_visual_feature_sequence) + ((1 - visual_gate_values) * features_text_visual_interaction)
        # fused_text_dominant = (text_gate_values * text_feature_sequence) + ((1 - text_gate_values) * features_visual_text_interaction)
        # Conceptual final fusion (depends on conceptual gated fusions)
        # calculated_fused_shared_features = (self.fusion_weighting_alpha1 * fused_representation_1 + ...)
        # --- End of Conceptual Shared Fusion Calculation Section ---

        # --- Step 4: Original Multimodal Feature Combination (Simple Concatenation - FUNCTIONALLY USED) ---
        # *** THIS IS THE KEY STEP MAINTAINING ORIGINAL FUNCTIONALITY ***
        concatenated_encoder_features = torch.cat([text_feature_sequence, projected_visual_feature_sequence], dim=1)

        # --- Step 5: Prepare Extended Attention Mask for Concatenated Features ---
        batch_size, text_sequence_length = encoder_input_token_ids.shape
        visual_sequence_length = projected_visual_feature_sequence.shape[1]
        text_part_attention_mask = encoder_attention_mask_tensor
        visual_part_attention_mask = torch.ones((batch_size, visual_sequence_length), dtype=text_part_attention_mask.dtype, device=text_part_attention_mask.device)
        extended_concatenated_attention_mask = torch.cat([text_part_attention_mask, visual_part_attention_mask], dim=1)

        # --- Step 6: Pass Combined Features to BART Decoder for Generation/Loss ---
        final_model_outputs = self.language_model_generator_decoder(
            encoder_outputs=BaseModelOutput(last_hidden_state=concatenated_encoder_features),
            attention_mask=extended_concatenated_attention_mask,
            decoder_input_ids=decoder_input_ids,
            labels=labels,
            return_dict=True
        )
        return final_model_outputs
print("AdvancedSarcasmExplanationModel class defined.")


# %% Image Data Organization (Copying to working directory for accessibility)
print("\n--- Organizing Image Files ---")
# Define paths for source images and destination working directories
original_image_dataset_directory = "/kaggle/input/more-plus/MORE-PLUS-DATASET/images"
working_directory_train_images = "/kaggle/working/images/train"
working_directory_val_images = "/kaggle/working/images/val"

# Create the destination directories in the working space if they don't already exist
print(f"Creating working directory for training images: {working_directory_train_images}")
os.makedirs(working_directory_train_images, exist_ok=True)
print(f"Creating working directory for validation images: {working_directory_val_images}")
os.makedirs(working_directory_val_images, exist_ok=True)

# Function to copy image files based on the provided dataframe PIDs
def organize_image_files_by_split(source_pid_dataframe, destination_image_directory, source_images_root_path):
    """Copies images corresponding to PIDs in the dataframe to the destination directory."""
    print(f"Starting image organization process for: {destination_image_directory}")
    copied_count = 0
    skipped_count = 0
    # Iterate through each post ID in the dataframe
    for post_identifier in tqdm(source_pid_dataframe['pid'], desc=f"Copying images to {os.path.basename(destination_image_directory)}", leave=False):
        post_id_string = str(post_identifier).strip() # Ensure PID is treated as a string
        source_file_path = os.path.join(source_images_root_path, f"{post_id_string}.jpg")
        destination_file_path = os.path.join(destination_image_directory, f"{post_id_string}.jpg")

        # Check if the source file exists before attempting to copy
        if os.path.exists(source_file_path):
            # Avoid re-copying if it already exists in the destination
            if not os.path.exists(destination_file_path):
                 shutil.copy(source_file_path, destination_file_path)
                 copied_count += 1
            else:
                 skipped_count +=1

        else:
            # This case should be prevented by the earlier filtering
            print(f"Warning: Source image file not found during copy: {source_file_path}. This should not happen after filtering.")
    print(f"Image organization for {os.path.basename(destination_image_directory)} complete. Copied: {copied_count}, Skipped (already exist): {skipped_count}")


# Execute the image copying process for both training and validation sets
organize_image_files_by_split(training_dataframe, working_directory_train_images, original_image_dataset_directory)
organize_image_files_by_split(validation_dataframe, working_directory_val_images, original_image_dataset_directory)
print("Image file organization into working directories is complete.")


# %% Dataset and DataLoader Preparation Stage
print("\n--- Preparing Datasets and DataLoaders ---")
# Initialize the BART tokenizer
print("Initializing BART Tokenizer...")
text_sequence_tokenizer = BartTokenizer.from_pretrained("facebook/bart-base")

# Image transformations are defined within the Dataset class, using defaults here.

# Instantiate the training dataset using the custom Dataset class
print("Creating Training Dataset instance...")
training_dataset_instance = MultimodalSarcasmDataset(
    input_dataframe=training_dataframe,
    description_dictionary=description_data_train,
    object_dictionary=object_data_train,
    dataset_image_directory_path=working_directory_train_images,
    sequence_tokenizer_instance=text_sequence_tokenizer,
    image_transformations_pipeline=None # Use default transformations
)

# Instantiate the validation dataset using the custom Dataset class
print("Creating Validation Dataset instance...")
validation_dataset_instance = MultimodalSarcasmDataset(
    input_dataframe=validation_dataframe,
    description_dictionary=description_data_val,
    object_dictionary=object_data_val,
    dataset_image_directory_path=working_directory_val_images,
    sequence_tokenizer_instance=text_sequence_tokenizer,
    image_transformations_pipeline=None # Use default transformations
)

# Define data loader parameters (batch size, number of workers for parallel loading)
dataloader_batch_size = 8 # Adjust based on GPU memory
dataloader_num_workers = 2 # Adjust based on system cores and memory

# Create the training data loader instance
print(f"Creating Training DataLoader (Batch Size: {dataloader_batch_size}, Workers: {dataloader_num_workers})...")
training_data_loader = DataLoader(
    training_dataset_instance,
    batch_size=dataloader_batch_size,
    shuffle=True, # Shuffle training data every epoch for better generalization
    num_workers=dataloader_num_workers,
    pin_memory=True, # Speeds up data transfer to GPU by pinning memory
    # collate_fn=None # Use default collate function
    drop_last=True # Drop last incomplete batch if dataset size is not divisible by batch size
)

# Create the validation data loader instance
print(f"Creating Validation DataLoader (Batch Size: {dataloader_batch_size}, Workers: {dataloader_num_workers})...")
validation_data_loader = DataLoader(
    validation_dataset_instance,
    batch_size=dataloader_batch_size,
    shuffle=False, # No need to shuffle validation data
    num_workers=dataloader_num_workers,
    pin_memory=True,
    # collate_fn=None
    drop_last=False # Keep last incomplete batch for validation
)

print("Dataset and DataLoader setup is complete.")

# %% Model Training Configuration and Execution Loop
print("\n--- Model Training ---")
# Determine the computation device (GPU if available, otherwise CPU)
computation_device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Selected computation device for training: {computation_device}")

# Initialize the advanced multimodal sarcasm explanation model and move it to the selected device
print("Initializing the AdvancedSarcasmExplanationModel...")
sarcasm_explanation_model = AdvancedSarcasmExplanationModel().to(computation_device)

# --- Optimizer Configuration ---
# Using AdamW optimizer (Adam with Weight Decay), suitable for Transformer models
print("Configuring AdamW optimizer...")
model_optimizer = AdamW(sarcasm_explanation_model.parameters(), lr=5e-5, weight_decay=0.01)

# --- Training Hyperparameters ---
number_of_training_epochs = 3
# Calculate total number of training steps for the learning rate scheduler
# Use len(training_data_loader) which accounts for drop_last=True
total_training_steps_count = number_of_training_epochs * len(training_data_loader)
print(f"Training Configuration: Epochs={number_of_training_epochs}, Steps per Epoch={len(training_data_loader)}, Total Steps={total_training_steps_count}")

# --- Learning Rate Scheduler ---
# Using a linear decay scheduler from Hugging Face's transformers library
print("Configuring linear learning rate scheduler...")
learning_rate_scheduler = get_scheduler(
    name="linear",                     # Type of scheduler
    optimizer=model_optimizer,         # The optimizer to schedule
    num_warmup_steps=0,                # Number of warmup steps (0 means no warmup)
    num_training_steps=total_training_steps_count, # Total number of training steps
)

# --- Checkpoint Saving Configuration ---
# Define a filename template for saving model checkpoints after each epoch
model_checkpoint_save_path_template = "sarcasm_explainer_epoch_{}.pt"
print(f"Model checkpoints will be saved using template: {model_checkpoint_save_path_template}")

# --- Training History ---
training_history = {'epoch': [], 'train_loss': [], 'val_loss': []}

print("\nStarting the model training loop...")
# --- Model Training Loop ---
for current_epoch_number in range(number_of_training_epochs):
    epoch_str = f"Epoch {current_epoch_number+1}/{number_of_training_epochs}"
    print(f"\n--- {epoch_str} ---")

    # --- Training Phase ---
    sarcasm_explanation_model.train() # Set model to training mode
    cumulative_epoch_training_loss = 0.0
    training_loop_progress_bar = tqdm(training_data_loader, desc=f"{epoch_str} [Training]", leave=False)

    for batch_index, batch_data_dict in enumerate(training_loop_progress_bar):
        # Move tensors to device
        encoder_input_ids_batch = batch_data_dict['encoder_input_token_ids'].to(computation_device)
        encoder_attention_mask_batch = batch_data_dict['encoder_attention_mask_tensor'].to(computation_device)
        image_data_batch = batch_data_dict['image_data_tensor'].to(computation_device)
        decoder_input_ids_batch = batch_data_dict['decoder_target_token_ids'][:, :-1].contiguous().to(computation_device)
        target_labels_batch = batch_data_dict['decoder_target_token_ids'][:, 1:].contiguous().to(computation_device)
        target_labels_batch[target_labels_batch == text_sequence_tokenizer.pad_token_id] = -100

        # Forward pass
        model_outputs_dict = sarcasm_explanation_model(
            encoder_input_token_ids=encoder_input_ids_batch,
            encoder_attention_mask_tensor=encoder_attention_mask_batch,
            image_data_tensor=image_data_batch,
            decoder_input_ids=decoder_input_ids_batch,
            labels=target_labels_batch
        )
        current_batch_loss = model_outputs_dict.loss

        # Backward pass & optimization
        model_optimizer.zero_grad()
        current_batch_loss.backward()
        torch.nn.utils.clip_grad_norm_(sarcasm_explanation_model.parameters(), max_norm=1.0) # Gradient clipping
        model_optimizer.step()
        learning_rate_scheduler.step()

        cumulative_epoch_training_loss += current_batch_loss.item()
        training_loop_progress_bar.set_postfix(batch_loss=f"{current_batch_loss.item():.4f}")

    average_epoch_training_loss = cumulative_epoch_training_loss / len(training_data_loader)

    # --- Validation Phase ---
    sarcasm_explanation_model.eval() # Set model to evaluation mode
    cumulative_epoch_validation_loss = 0.0
    validation_loop_progress_bar = tqdm(validation_data_loader, desc=f"{epoch_str} [Validation]", leave=False)

    with torch.no_grad():
        for batch_data_dict in validation_loop_progress_bar:
            # Move tensors to device
            encoder_input_ids_batch = batch_data_dict['encoder_input_token_ids'].to(computation_device)
            encoder_attention_mask_batch = batch_data_dict['encoder_attention_mask_tensor'].to(computation_device)
            image_data_batch = batch_data_dict['image_data_tensor'].to(computation_device)
            decoder_input_ids_batch = batch_data_dict['decoder_target_token_ids'][:, :-1].contiguous().to(computation_device)
            target_labels_batch = batch_data_dict['decoder_target_token_ids'][:, 1:].contiguous().to(computation_device)
            target_labels_batch[target_labels_batch == text_sequence_tokenizer.pad_token_id] = -100

            # Forward pass
            model_outputs_dict = sarcasm_explanation_model(
                encoder_input_token_ids=encoder_input_ids_batch,
                encoder_attention_mask_tensor=encoder_attention_mask_batch,
                image_data_tensor=image_data_batch,
                decoder_input_ids=decoder_input_ids_batch,
                labels=target_labels_batch
            )
            current_batch_loss = model_outputs_dict.loss
            cumulative_epoch_validation_loss += current_batch_loss.item()
            validation_loop_progress_bar.set_postfix(batch_loss=f"{current_batch_loss.item():.4f}")

    average_epoch_validation_loss = cumulative_epoch_validation_loss / len(validation_data_loader)

    # Print loss summary & store history
    print(f"{epoch_str} Summary:")
    print(f"  Average Training Loss: {average_epoch_training_loss:.4f}")
    print(f"  Average Validation Loss: {average_epoch_validation_loss:.4f}")
    training_history['epoch'].append(current_epoch_number + 1)
    training_history['train_loss'].append(average_epoch_training_loss)
    training_history['val_loss'].append(average_epoch_validation_loss)


    # --- Save Model Checkpoint After Each Epoch ---
    current_epoch_checkpoint_path = model_checkpoint_save_path_template.format(current_epoch_number + 1)
    torch.save(sarcasm_explanation_model.state_dict(), current_epoch_checkpoint_path)
    print(f"  Model checkpoint saved: {current_epoch_checkpoint_path}")

print("\n--- Model training loop finished. ---")

# Plot training history
print("\nPlotting training and validation loss history...")
plt.figure(figsize=(10, 5))
plt.plot(training_history['epoch'], training_history['train_loss'], label='Training Loss')
plt.plot(training_history['epoch'], training_history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss Over Epochs')
plt.legend()
plt.grid(True)
plt.show()


# %% Model Evaluation Phase
print("\n--- Model Evaluation ---")
# --- Load the Model Checkpoint for Evaluation ---
# Load the model from the *last* epoch as requested
final_epoch_checkpoint_path = model_checkpoint_save_path_template.format(number_of_training_epochs)
print(f"Loading model state dictionary for evaluation from: {final_epoch_checkpoint_path}")

# Check if the checkpoint file exists
if not os.path.exists(final_epoch_checkpoint_path):
     raise FileNotFoundError(f"Checkpoint file not found: {final_epoch_checkpoint_path}. Training might have failed or saving path is incorrect.")

# Re-initialize the model structure first
evaluation_model_instance = AdvancedSarcasmExplanationModel().to(computation_device) # Ensure model is on the correct device
# Load the saved state dictionary into the model instance
evaluation_model_instance.load_state_dict(torch.load(final_epoch_checkpoint_path, map_location=computation_device))
evaluation_model_instance.eval() # Set the loaded model to evaluation mode
print("Model loaded successfully and set to evaluation mode.")

# --- Prepare Lists to Store Generation Results ---
list_of_predicted_explanation_texts = []
list_of_ground_truth_explanation_texts = []
list_of_original_input_texts = []

# --- Generate Predictions on the Validation Set ---
print("Generating sarcasm explanations for the validation dataset...")
with torch.no_grad(): # Ensure gradient calculations are off during inference
    # Iterate through the validation data loader
    for batch_data_dict in tqdm(validation_data_loader, desc="Generating Predictions", leave=False):
        # Move necessary input tensors to the computation device
        encoder_input_ids_batch = batch_data_dict['encoder_input_token_ids'].to(computation_device)
        encoder_attention_mask_batch = batch_data_dict['encoder_attention_mask_tensor'].to(computation_device)
        image_data_batch = batch_data_dict['image_data_tensor'].to(computation_device)

        # --- Replicate Feature Extraction and Concatenation Logic ---
        visual_feature_outputs = evaluation_model_instance.image_feature_extractor_model(pixel_values=image_data_batch)
        raw_visual_feature_sequence = visual_feature_outputs.last_hidden_state
        projected_visual_feature_sequence = evaluation_model_instance.image_feature_projection_layer(raw_visual_feature_sequence)
        text_encoder_component = evaluation_model_instance.language_model_generator_decoder.model.encoder
        text_feature_outputs = text_encoder_component(input_ids=encoder_input_ids_batch, attention_mask=encoder_attention_mask_batch)
        text_feature_sequence = text_feature_outputs.last_hidden_state
        concatenated_encoder_features = torch.cat([text_feature_sequence, projected_visual_feature_sequence], dim=1)
        batch_size_eval, text_seq_len_eval = encoder_input_ids_batch.shape
        visual_seq_len_eval = projected_visual_feature_sequence.shape[1]
        visual_part_mask_eval = torch.ones((batch_size_eval, visual_seq_len_eval), dtype=encoder_attention_mask_batch.dtype, device=computation_device)
        extended_concatenated_attention_mask = torch.cat([encoder_attention_mask_batch, visual_part_mask_eval], dim=1)
        encoder_outputs_wrapper_for_generation = BaseModelOutput(last_hidden_state=concatenated_encoder_features)

        # --- Generate Text using BART's `generate` Method ---
        generated_token_ids = evaluation_model_instance.language_model_generator_decoder.generate(
            encoder_outputs=encoder_outputs_wrapper_for_generation,
            attention_mask=extended_concatenated_attention_mask,
            max_length=64,
            num_beams=4,
            early_stopping=True,
            no_repeat_ngram_size=2 # Added to reduce repetition
        )

        # --- Decode Generated Token IDs and Reference Token IDs to Text ---
        predicted_batch_texts_list = text_sequence_tokenizer.batch_decode(generated_token_ids, skip_special_tokens=True)
        actual_batch_texts_list = text_sequence_tokenizer.batch_decode(batch_data_dict['decoder_target_token_ids'], skip_special_tokens=True)

        # --- Store Decoded Texts ---
        list_of_predicted_explanation_texts.extend(predicted_batch_texts_list)
        list_of_ground_truth_explanation_texts.extend(actual_batch_texts_list)
        list_of_original_input_texts.extend(batch_data_dict['raw_input_text_string'])

# --- Display Sample Predictions for Qualitative Assessment ---
print("\n=== Sample Generated vs. Reference Explanations ===")
number_of_samples_to_display = min(5, len(list_of_predicted_explanation_texts))
for i in range(number_of_samples_to_display):
    print(f"--- Sample #{i+1} ---")
    print(f"Original Input Text: {list_of_original_input_texts[i]}")
    print(f"Reference (GT) Explanation: {list_of_ground_truth_explanation_texts[i]}")
    print(f"Generated Explanation: {list_of_predicted_explanation_texts[i]}")
    print("-" * 90)

# %% Calculate Standard Evaluation Metrics for Text Generation
print("\n--- Calculating Evaluation Metrics ---")
# --- Load Evaluation Metric Calculators ---
print("Loading standard text generation evaluation metrics (ROUGE, BLEU, METEOR, BERTScore)...")
rouge_metric_calculator = load("rouge")
bleu_metric_calculator = load("bleu")
meteor_metric_calculator = load("meteor")
bertscore_metric_calculator = load("bertscore")
print("Evaluation metric calculators loaded.")

# Ensure predictions and references are not empty before calculating metrics
if not list_of_predicted_explanation_texts or not list_of_ground_truth_explanation_texts:
    print("Error: No predictions or references found. Skipping metric calculation.")
else:
    # --- Compute ROUGE Scores ---
    print("Calculating ROUGE scores...")
    rouge_evaluation_results = rouge_metric_calculator.compute(
        predictions=list_of_predicted_explanation_texts,
        references=list_of_ground_truth_explanation_texts
    )

    # --- Compute BLEU Score ---
    print("Calculating BLEU score...")
    bleu_references_formatted = [[ref] for ref in list_of_ground_truth_explanation_texts]
    bleu_evaluation_result = bleu_metric_calculator.compute(
        predictions=list_of_predicted_explanation_texts,
        references=bleu_references_formatted
    )

    # --- Compute METEOR Score ---
    print("Calculating METEOR score...")
    meteor_evaluation_result = meteor_metric_calculator.compute(
        predictions=list_of_predicted_explanation_texts,
        references=list_of_ground_truth_explanation_texts
    )

    # --- Compute BERTScore ---
    print("Calculating BERTScore (this may take a moment)...")
    try:
        bertscore_evaluation_results_dict = bertscore_metric_calculator.compute(
            predictions=list_of_predicted_explanation_texts,
            references=list_of_ground_truth_explanation_texts,
            lang="en",
            model_type="bert-base-uncased",
            device=computation_device # Explicitly use the computation device
        )
        average_bertscore_f1_value = np.mean(bertscore_evaluation_results_dict['f1'])
        print("Metric calculations complete.")

        # --- Print Consolidated Evaluation Results ---
        print("\n=== Consolidated Quantitative Evaluation Results ===")
        print(f"ROUGE Scores: {rouge_evaluation_results}")
        print(f"BLEU Score: {bleu_evaluation_result}")
        print(f"METEOR Score: {meteor_evaluation_result}")
        print(f"BERTScore (Average F1): {average_bertscore_f1_value:.4f}")

    except Exception as e:
        print(f"Error calculating BERTScore: {e}")
        print("Skipping BERTScore calculation.")
        # Print other results if available
        print("\n=== Partial Quantitative Evaluation Results ===")
        if 'rouge_evaluation_results' in locals(): print(f"ROUGE Scores: {rouge_evaluation_results}")
        if 'bleu_evaluation_result' in locals(): print(f"BLEU Score: {bleu_evaluation_result}")
        if 'meteor_evaluation_result' in locals(): print(f"METEOR Score: {meteor_evaluation_result}")


# %% Save Prediction Results to File
print("\n--- Saving Prediction Results ---")
# --- Create a Pandas DataFrame ---
number_of_predictions_made = len(list_of_predicted_explanation_texts)
# Adjust slice for validation_dataframe based on actual predictions made
validation_ids_slice = validation_dataframe["pid"][:number_of_predictions_made].tolist()

if len(validation_ids_slice) == number_of_predictions_made and \
   len(list_of_original_input_texts) == number_of_predictions_made and \
   len(list_of_ground_truth_explanation_texts) == number_of_predictions_made:

    prediction_results_dataframe = pd.DataFrame({
        "post_id": validation_ids_slice,
        "original_text": list_of_original_input_texts,
        "reference_explanation": list_of_ground_truth_explanation_texts,
        "generated_explanation": list_of_predicted_explanation_texts
    })

    # --- Save the DataFrame to a CSV file ---
    output_csv_filename = "detailed_sarcasm_explanation_predictions.csv"
    print(f"\nSaving prediction results DataFrame to: {output_csv_filename}")
    prediction_results_dataframe.to_csv(output_csv_filename, index=False)
    print("Prediction results saved successfully.")
else:
    print(f"Warning: Length mismatch when creating results DataFrame. Skipping save.")
    print(f"  - Expected length: {number_of_predictions_made}")
    print(f"  - PIDs: {len(validation_ids_slice)}")
    print(f"  - Original Texts: {len(list_of_original_input_texts)}")
    print(f"  - GT Explanations: {len(list_of_ground_truth_explanation_texts)}")


# %% Function for Testing on a Single Instance
print("\n--- Defining Single Instance Test Function ---")

def test_single_instance(input_text, image_path, model_checkpoint_path, tokenizer, device):
    """
    Loads the trained model and generates an explanation for a single
    text-image pair.

    Args:
        input_text (str): The sarcastic text input.
        image_path (str): Path to the input image file.
        model_checkpoint_path (str): Path to the saved model state dictionary (.pt file).
        tokenizer: The loaded BART tokenizer instance.
        device: The torch device (e.g., 'cuda' or 'cpu').

    Returns:
        str: The generated explanation string.
    """
    print(f"\n--- Testing Single Instance ---")
    print(f"Input Text: {input_text}")
    print(f"Image Path: {image_path}")
    print(f"Model Checkpoint: {model_checkpoint_path}")

    # --- 1. Load Model ---
    if not os.path.exists(model_checkpoint_path):
        print(f"Error: Model checkpoint not found at {model_checkpoint_path}")
        return "[Error: Model checkpoint not found]"

    print("Loading model for inference...")
    inference_model = AdvancedSarcasmExplanationModel().to(device)
    try:
        inference_model.load_state_dict(torch.load(model_checkpoint_path, map_location=device))
        inference_model.eval()
        print("Model loaded successfully.")
    except Exception as e:
        print(f"Error loading model state_dict: {e}")
        return f"[Error: Could not load model from {model_checkpoint_path}]"


    # --- 2. Preprocess Image ---
    print("Processing image...")
    try:
        image_transforms = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        image = Image.open(image_path).convert('RGB')
        image_tensor = image_transforms(image).unsqueeze(0).to(device) # Add batch dimension and move to device
    except FileNotFoundError:
        print(f"Error: Image file not found at {image_path}")
        return "[Error: Image file not found]"
    except Exception as e:
        print(f"Error processing image: {e}")
        return "[Error: Image processing failed]"

    # --- 3. Prepare Text Input ---
    print("Preparing text input...")
    # Use placeholders for missing context (description, objects, target)
    # In a real application, you might fetch/generate these if possible.
    context_description_text = ""  # Placeholder
    context_objects_string = ""    # Placeholder
    sarcasm_focus_target_info = "" # Placeholder
    combined_input_sequence_text = f"{input_text} [DESC] {context_description_text} [OBJ] {context_objects_string} [TARGET] {sarcasm_focus_target_info}"

    tokenized_encoder_input = tokenizer(
        combined_input_sequence_text,
        return_tensors='pt',
        padding='max_length',
        truncation=True,
        max_length=256
    )
    encoder_input_ids = tokenized_encoder_input.input_ids.to(device)
    encoder_attention_mask = tokenized_encoder_input.attention_mask.to(device)

    # --- 4. Perform Inference ---
    print("Generating explanation...")
    with torch.no_grad():
        # Replicate feature extraction and concatenation logic
        visual_feature_outputs = inference_model.image_feature_extractor_model(pixel_values=image_tensor)
        raw_visual_feature_sequence = visual_feature_outputs.last_hidden_state
        projected_visual_feature_sequence = inference_model.image_feature_projection_layer(raw_visual_feature_sequence)

        text_encoder_component = inference_model.language_model_generator_decoder.model.encoder
        text_feature_outputs = text_encoder_component(input_ids=encoder_input_ids, attention_mask=encoder_attention_mask)
        text_feature_sequence = text_feature_outputs.last_hidden_state

        concatenated_encoder_features = torch.cat([text_feature_sequence, projected_visual_feature_sequence], dim=1)

        batch_size_inf, text_seq_len_inf = encoder_input_ids.shape
        visual_seq_len_inf = projected_visual_feature_sequence.shape[1]
        visual_part_mask_inf = torch.ones((batch_size_inf, visual_seq_len_inf), dtype=encoder_attention_mask.dtype, device=device)
        extended_concatenated_attention_mask = torch.cat([encoder_attention_mask, visual_part_mask_inf], dim=1)

        encoder_outputs_wrapper_for_generation = BaseModelOutput(last_hidden_state=concatenated_encoder_features)

        # Generate using the model
        generated_ids = inference_model.language_model_generator_decoder.generate(
            encoder_outputs=encoder_outputs_wrapper_for_generation,
            attention_mask=extended_concatenated_attention_mask,
            max_length=64,
            num_beams=4,
            early_stopping=True,
            no_repeat_ngram_size=2
        )

    # --- 5. Decode Output ---
    generated_explanation = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
    print("Explanation generated.")

    return generated_explanation

print("Single instance test function defined.")


# %% Example Usage of the Test Function
print("\n--- Example: Testing the Model on a Single Instance ---")

# Define the path to the model checkpoint you want to test (e.g., the last epoch)
test_model_checkpoint = model_checkpoint_save_path_template.format(number_of_training_epochs)

# Define example text and the path to an image file
# **IMPORTANT**: Replace 'path/to/your/test_image.jpg' with an actual image path accessible by the notebook.
# You might need to upload an image or use one from the validation set copied earlier.
example_text = "Oh great, another meeting that could have been an email."
example_image_path = os.path.join(working_directory_val_images, validation_dataframe.iloc[0]['pid'] + ".jpg") # Use first val image path

# Check if the example image exists before testing
if os.path.exists(example_image_path) and os.path.exists(test_model_checkpoint):
    # Ensure the tokenizer used here is the same one used for training
    test_tokenizer = BartTokenizer.from_pretrained("facebook/bart-base")
    test_device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Call the test function
    generated_explanation = test_single_instance(
        input_text=example_text,
        image_path=example_image_path,
        model_checkpoint_path=test_model_checkpoint,
        tokenizer=test_tokenizer,
        device=test_device
    )

    print("\n--- Generated Explanation for Example ---")
    print(f"Input Text: {example_text}")
    print(f"Image Used: {example_image_path}")
    print(f"Generated Explanation: {generated_explanation}")

elif not os.path.exists(test_model_checkpoint):
     print(f"Skipping single instance test: Model checkpoint not found at {test_model_checkpoint}")
elif not os.path.exists(example_image_path):
     print(f"Skipping single instance test: Example image not found at {example_image_path}. Please provide a valid path.")


print("\n=== Full Script Execution Finished ===")
```
